```python
import requests

url = "https://fantasy.premierleague.com/api/bootstrap-static/"

response = requests.get(url)

data = response.json()

data.keys()
```

```python
# 1. player id

identity_fields = [
    "id",
    "first_name",
    "second_name",
    "web_name"
]

for field in identity_fields:
    print(f"{field}: {data['elements'][0][field]}")
```

```python
# 2.teams and position

team_position_fields = [
    "team",
    "element_type"
]

for field in team_position_fields:
    print(f"{field}: {data['elements'][0][field]}")
```

```python
# 3. Price & ownership

price = data["elements"][0]["now_cost"] / 10
ownership = data["elements"][0]["selected_by_percent"]

print(f"now_cost: £{price:.1f}m")
print(f"selected_by_percent: {ownership}%")
```

```python
# 4. FPL performance


performance_fields = [
    "total_points",
    "form",
    "points_per_game",
    "minutes",
    "goals_scored",
    "assists",
    "clean_sheets"
]

for field in performance_fields:
    print(f"{field}: {data['elements'][0][field]}")
```

```python
# 5. Expected statistics

expected_stats_fields = [
    "expected_goals",
    "expected_assists",
    "expected_goal_involvements"
]

for field in expected_stats_fields:
    print(f"{field}: {data['elements'][0][field]}")
```

```python
# 6. ICT metrics


ict_fields = [
    "ict_index",
    "influence",
    "creativity",
    "threat"
]

for field in ict_fields:
    print(f"{field}: {data['elements'][0][field]}")
```

```python
#7. Availability


availability_fields = [
    "status",
    "chance_of_playing_next_round"
]

for field in availability_fields:
    print(f"{field}: {data['elements'][0][field]}")
```

```python
# 8. Transfers


transfer_fields = [
    "transfers_in_event",
    "transfers_out_event"
]

for field in transfer_fields:
    print(f"{field}: {data['elements'][0][field]}")
```

```python
#9. Price changes


price_change_fields = [
    "cost_change_start",
    "cost_change_event"
]

for field in price_change_fields:
    print(f"{field}: {data['elements'][0][field]}")
```
