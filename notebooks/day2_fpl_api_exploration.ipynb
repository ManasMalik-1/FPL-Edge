### fct_player_gameweek

Grain: **One row per player per gameweek.**

This table stores a player's gameweek-level performance data, including:

- Total points
- Minutes played
- Goals scored
- Assists
- Clean sheets
- Bonus points
- BPS
- Expected goals (xG)
- Expected assists (xA)
- Expected goal involvements (xGI)
- ICT statistics
- Historical player price
